## 1. Overview & Intuition

Traditional search matches exact keywords (Lexical Search). Modern AI uses **Vector Embeddings** to convert sentences into coordinates in a multi-dimensional geometry space (Semantic Search). Words and documents with similar meanings land close to each other.

## 2. Building a Semantic Search Engine

```python
from sentence_transformers import SentenceTransformer
import numpy as np

# 1. Load lightweight embedding model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Document Corpus
documents = [
    "A quick brown fox jumps over the lazy dog.",
    "Python is an interpreted, high-level programming language used for machine learning.",
    "The recipe requires two cups of flour, sugar, and organic butter.",
    "Vehicles powered by lithium-ion batteries produce zero tailpipe emissions.",
    "Artificial neural networks are computational systems inspired by biological brains.",
    "Baking chocolate chip cookies in a preheated oven at 350 degrees."
]

doc_embeddings = model.encode(documents, convert_to_numpy=True)

# 3. Cosine similarity function
def search(query, top_k=2):
    q_emb = model.encode(query, convert_to_numpy=True)
    scores = np.dot(doc_embeddings, q_emb) / (
        np.linalg.norm(doc_embeddings, axis=1) * np.linalg.norm(q_emb)
    )
    ranked_indices = np.argsort(scores)[::-1][:top_k]
    
    print(f"\nSearch Query: '{query}'")
    for rank, idx in enumerate(ranked_indices, 1):
        print(f"{rank}. [Score: {scores[idx]:.4f}] {documents[idx]}")

# Run test queries with zero keyword matches
search("electric automobiles and clean transit")
search("sweet dessert recipes")
search("coding intelligent computer algorithms")
```
