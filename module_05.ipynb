## 1. Overview

Generative Large Language Models (LLMs) operate via autoregression: predicting the next most probable token given the context so far. The **Transformer** uses **Self-Attention** to dynamically look at other words across the sequence simultaneously.

## 2. Text Generation and Sampling (Temperature Control)

```python
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

# 1. Load lightweight open model
model_name = "gpt2"
print(f"Loading {model_name}...")
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
model.eval()

# 2. Generation function with adjustable temperature
def generate_text(prompt, temp=0.7, max_tokens=35):
    inputs = tokenizer(prompt, return_tensors="pt")
    with torch.no_grad():
        out = model.generate(
            inputs.input_ids,
            max_new_tokens=max_tokens,
            do_sample=True,
            temperature=temp,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id
        )
    return tokenizer.decode(out[0], skip_special_tokens=True)

test_prompt = "The key breakthroughs in artificial intelligence will come from"

print("=== Strict / Deterministic (Temperature = 0.2) ===")
print(generate_text(test_prompt, temp=0.2))

print("\n=== Creative / Varied (Temperature = 0.9) ===")
print(generate_text(test_prompt, temp=0.9))
```
